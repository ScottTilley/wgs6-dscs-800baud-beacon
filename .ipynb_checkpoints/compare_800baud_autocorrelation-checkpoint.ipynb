# 800-baud beacon autocorrelation comparison

Overlay comparison of **USA-134**, **USA-167**, and **WGS-6** using the same basic framing/autocorrelation approach as the Dani notebook.

The notebook:
- reads the three float32 soft-symbol files,
- finds the best 8-bit serial frame offset independently for each,
- resolves BPSK polarity using frame bit 0,
- computes normalized autocorrelation,
- overlays all three satellites in the **same plot area**.


```python
import numpy as np
import matplotlib.pyplot as plt
import scipy.signal
from pathlib import Path
```

## Input files

```python
files = {
    'USA-134': Path('symbols_800baud_usa134.f32'),
    'USA-167': Path('symbols_800baud_usa167.f32'),
    'WGS-6':   Path('symbols_800baud_wgs6.f32'),
}

for label, path in files.items():
    print(f'{label:7s}: {path}  exists={path.exists()}')
```

## Frame alignment and polarity correction

```python
def load_and_frame(path):
    x = np.fromfile(path, dtype=np.float32)

    offset_scores = []
    for off in range(8):
        usable = ((x.size - off) // 8) * 8
        z = x[off:off + usable].reshape((-1, 8))
        offset_scores.append(abs(float(np.mean(z[:, 0]))))

    off = int(np.argmax(offset_scores))
    usable = ((x.size - off) // 8) * 8
    z = x[off:off + usable].reshape((-1, 8))

    inverted = False
    if np.mean(z[:, 0]) > 0:
        z = -z
        inverted = True

    return x, z, off, inverted, offset_scores

framed = {}

for label, path in files.items():
    x, z, off, inverted, scores = load_and_frame(path)
    framed[label] = z
    print(
        f'{label:7s}: {x.size:7d} symbols, {z.shape[0]:6d} frames, '
        f'offset={off}, inverted={inverted}'
    )
```

## Normalized autocorrelation helper

```python
def normalized_autocorr(v, max_lag=100):
    v = np.asarray(v, dtype=np.float64)
    v = v - np.mean(v)

    power = np.dot(v, v)
    if power <= 1e-18:
        return np.ones(max_lag + 1)

    corr = scipy.signal.correlate(v, v, mode='full', method='fft')
    corr = corr[corr.size // 2:]
    corr = corr / corr[0]
    return corr[:max_lag + 1]
```

## Overlay comparison: aggregate of frame bits 1–5

Each satellite trace below is the mean of the normalized autocorrelation curves for frame bits 1 through 5. That keeps the comparison to exactly **three curves in one graph** while showing the common repeating structure.

```python
max_lag = 100
lags = np.arange(max_lag + 1)

plt.figure(figsize=(14, 8))

for label, z in framed.items():
    acfs = np.vstack([
        normalized_autocorr(z[:, bit], max_lag=max_lag)
        for bit in range(1, 6)
    ])
    aggregate = np.mean(acfs, axis=0)
    plt.plot(lags, aggregate, linewidth=2, label=label)

for p in [25, 27, 29, 31, 32]:
    plt.axvline(p, linestyle=':', linewidth=1, alpha=0.5)

plt.title('800-baud beacon autocorrelation comparison')
plt.xlabel('Lag (8-bit frames)')
plt.ylabel('Mean normalized autocorrelation, frame bits 1–5')
plt.xlim(0, max_lag)
plt.grid(True, alpha=0.25)
plt.legend(title='Satellite')
plt.tight_layout()
plt.show()
```

## Overlay one specific frame-bit channel

Change `bit_to_compare` from 1 through 5 to compare one of the individual published-period channels directly.

```python
bit_to_compare = 3
max_lag = 100
lags = np.arange(max_lag + 1)

plt.figure(figsize=(14, 8))

for label, z in framed.items():
    ac = normalized_autocorr(z[:, bit_to_compare], max_lag=max_lag)
    plt.plot(lags, ac, linewidth=2, label=label)

plt.title(f'800-baud frame bit {bit_to_compare} autocorrelation comparison')
plt.xlabel('Lag (8-bit frames)')
plt.ylabel('Normalized autocorrelation')
plt.xlim(0, max_lag)
plt.grid(True, alpha=0.25)
plt.legend(title='Satellite')
plt.tight_layout()
plt.show()
```
